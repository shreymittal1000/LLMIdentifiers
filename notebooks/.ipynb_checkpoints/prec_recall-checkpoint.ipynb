```python
from collections import defaultdict
import json
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import os
import pandas as pd
import seaborn as sns
from sklearn.metrics import precision_score, recall_score, confusion_matrix
```

```python
def load_data(path):
    data = []
    for root, dirs, files in os.walk(path):
        for filename in files:
            if filename.endswith(".json"):
                file_path = os.path.join(root, filename)
                try:
                    with open(file_path, "r") as f:
                        entry = json.load(f)
                        data.append(entry)
                except json.JSONDecodeError as e:
                    print(f"Skipped invalid JSON: {file_path} ({e})")
                    
    print(f"Loaded {len(data)} valid JSON result files from '{folder_path}' and its subfolders.")
    return data

folder_path = "../results/base_cues_5/"
data_cues_5 = load_data(folder_path)
folder_path = "../results/base_cues_10/"
data_cues_10 = load_data(folder_path)
folder_path = "../results/base_none_5/"
data_none_5 = load_data(folder_path)
folder_path = "../results/base_none_10/"
data_none_10 = load_data(folder_path)

data_cues = data_cues_5 + data_cues_10
data_none = data_none_5 + data_none_10
data_5 = data_cues_5 + data_none_5
data_10 = data_cues_10 + data_none_10
```

```python
from sklearn.metrics import confusion_matrix
import pandas as pd

labels = sorted(["gpt", "llama", "gemini", "claude", "qwen", "deepseek", "mistral"])
pretty_names = {
    "gpt": "GPT",
    "llama": "LLaMA",
    "mistral": "Mistral",
    "claude": "Claude",
    "qwen": "Qwen",
    "gemini": "Gemini",
    "deepseek": "DeepSeek"
}

def extract_useful_information(data):
    y_true, y_pred, y_guesser = [], [], []
    for entry in data:
        y_true.append(entry['model_general_name_1'])  # agent_0 guess target
        y_pred.append(entry['agent_0_guess'])
        y_guesser.append(entry['model_general_name_0'])
        y_true.append(entry['model_general_name_0'])  # agent_1 guess target
        y_pred.append(entry['agent_1_guess'])
        y_guesser.append(entry['model_general_name_1'])
    
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    y_guesser = np.array(y_guesser)
    y_true_pred = y_true == y_pred
    guesses_df = pd.DataFrame({"y_guesser": y_guesser, "y_guessee": y_true, "y_pred": y_pred, "y_pred_is_true": y_true_pred})
    return guesses_df

guesses_df_5c = extract_useful_information(data_cues_5)
guesses_df_10c = extract_useful_information(data_cues_10)
guesses_df_5n = extract_useful_information(data_none_5)
guesses_df_10n = extract_useful_information(data_none_10)

guesses_df_cues = extract_useful_information(data_cues)
guesses_df_none = extract_useful_information(data_none)
guesses_df_5 = extract_useful_information(data_5)
guesses_df_10 = extract_useful_information(data_10)
```

```python
def calculate(guesses_df):
    all_positives = guesses_df.groupby(["y_guesser", "y_pred"]).agg(all_positives=("y_pred", "size"))
    all_positives.index.set_names(['y_guesser', 'y_guessee'], inplace=True) 
    agg = guesses_df.groupby(["y_guesser", "y_guessee"]).agg(true_positives=("y_pred_is_true", "sum"), tp_and_fn=("y_guessee", "size"))
    agg["all_positives"] = all_positives["all_positives"]
    agg["precision"] = agg["true_positives"] / agg["all_positives"]
    agg["recall"] = agg["true_positives"] / agg["tp_and_fn"]
    agg.fillna(0.0, inplace=True)
    return agg

agg_5c = calculate(guesses_df_5c)
agg_10c = calculate(guesses_df_10c)
agg_5n = calculate(guesses_df_5n)
agg_10n = calculate(guesses_df_10n)

agg_c = calculate(guesses_df_cues)
agg_n = calculate(guesses_df_none)
agg_5 = calculate(guesses_df_5)
agg_10 = calculate(guesses_df_10)
```

```python
matplotlib.rcParams.update({"font.family": "serif"})

def plot_heatmap(df, save_file_name, title):
    matrix = df.unstack(level=1)
    plt.figure(figsize=(14, 12))
    ax = sns.heatmap(
        matrix,
        annot=True, fmt='.2f',
        cmap='YlGnBu',
        linewidths=0.3,
        cbar_kws={"shrink": 0.7},
        annot_kws={"fontsize": 24}
    )
    ax.set_title(title, fontsize=30)
    
    xticks = ax.get_xticklabels()
    yticks = ax.get_yticklabels()
    x_labels = [pretty_names.get(label.get_text(), label.get_text()) for label in xticks]
    y_labels = [pretty_names.get(label.get_text(), label.get_text()) for label in yticks]
    ax.set_xticklabels(x_labels, rotation=45, ha='right', fontsize=28)
    ax.set_yticklabels(y_labels, rotation=0, fontsize=28)
    
    ax.set_xlabel('Guessed Model', fontsize=28)
    ax.set_ylabel('Guesser Model', fontsize=28)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    cbar = ax.collections[0].colorbar
    cbar.ax.tick_params(labelsize=28)

    plt.tight_layout()
    plt.savefig("figures_and_files/" + save_file_name + ".eps", dpi=300, bbox_inches='tight')
    plt.show()
```

```python
plot_heatmap(agg_5c["precision"], "precision_c5", 'Precision by Model: 5 Rounds With Cues')
plot_heatmap(agg_5c["recall"], "recall_c5", 'Recall by Model: 5 Rounds With Cues')
print(agg_5c["precision"].unstack().to_latex())
print("==============")
print(agg_5c["recall"].unstack().to_latex())
```
