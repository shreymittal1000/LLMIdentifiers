````python
from collections import defaultdict
import json
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import os
import pandas as pd
import seaborn as sns
import concurrent.futures
from tqdm import tqdm
import math
import time, backoff
from openai import OpenAI
from dotenv import load_dotenv
from copy import deepcopy
import re
````

````python
def load_data(path):
    data = []
    for root, dirs, files in os.walk(path):
        for filename in files:
            if filename.endswith(".json"):
                file_path = os.path.join(root, filename)
                try:
                    with open(file_path, "r") as f:
                        entry = json.load(f)
                        data.append(entry)
                except json.JSONDecodeError as e:
                    print(f"Skipped invalid JSON: {file_path} ({e})")
                    
    print(f"Loaded {len(data)} valid JSON result files from '{folder_path}' and its subfolders.")
    return data

folder_path = "../results/base_cues_5/"
data_cues_5 = load_data(folder_path)
folder_path = "../results/base_cues_10/"
data_cues_10 = load_data(folder_path)

data_cues = data_cues_5 + data_cues_10
````

````python
load_dotenv()
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")

CUES = [
    "style-based utterance metrics (sentence length, bullet point usage, heading usage)",
    "TF-IDF",
    "Sentence Structure & Complexity",
    "Lexical Diversity & Vocabulary",
    "Punctuation & Formatting",
    "Grammar & Typographical Accuracy",
    "Formality & Tone",
    "Output Style & Structure",
    "Translation Performance & Style",
    "Emergent Model Characteristics",
]

class ORClient:
    def __init__(self, api_key, model="openai/gpt-4o-mini", min_interval=1.5):
        self.cli = OpenAI(api_key=api_key, base_url="https://openrouter.ai/api/v1")
        self.model = model
        self.min_interval = min_interval  # <- critical line
        self.last = 0

    @backoff.on_exception(backoff.expo, Exception, max_tries=6)
    def classify(self, prompt):
        wait = self.min_interval - (time.time() - self.last)
        if wait > 0:
            time.sleep(wait)
    
        res = self.cli.chat.completions.create(
            model=self.model,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=300,
            temperature=0,
        )
        self.last = time.time()
        text = res.choices[0].message.content.strip()
    
        # Try extracting the JSON object
        try:
            # Remove triple-backtick code block wrappers
            if text.startswith("```json"):
                text = text[7:].strip("` \n")
            elif text.startswith("```"):
                text = text[3:].strip("` \n")
    
            # Find first { ... }
            json_start = text.find("{")
            json_end = text.rfind("}") + 1
            json_str = text[json_start:json_end]
    
            return json.loads(json_str)

        except Exception as e:
            print("\n--- Failed to parse LLM output ---")
            print(text)
            print("----------------------------------\n")
            return {
                "Reasoning_behind_classification": f"Failed to parse: {text}",
                "Strategies": ["Failed to parse"]
            }
````

````python
def build_prompt(reasoning, guessed_model):
    allowed_str = "\n".join(f"{i+1}. {s}" for i, s in enumerate(CUES))
    return f"""
        You are judging which cues a language model used to guess another model, if any.

        The guessed model: **{guessed_model}**
        
        Classify the justification below into ANY of these strategies
        (you may output multiple, or "None" if none fit):
        
        {allowed_str}
        
        Justification:
        \"\"\"{reasoning}\"\"\"
        
        Return ONLY valid JSON like:
        {{  
          "Reasoning_behind_classification": "...explanation...",
          "Strategies": ["{CUES[1]}", "{CUES[3]}"]
        }}
    """

def extract_pairs(trials):
    pairs = []
    for t in trials:
        for agent_id in ["agent_0", "agent_1"]:
            reasoning = (t.get(f"{agent_id}_answer") or "").strip()
            guess = (t.get(f"{agent_id}_guess") or "").strip().lower()
            pairs.append({
                "guesser": agent_id,
                "guessed_model": guess,
                "reasoning": reasoning,
                "trial_id": t.get("seed", "na")
            })
    return pairs
````

````python
def run_classification(all_trials):
    client = ORClient(OPENROUTER_API_KEY)
    pairs = extract_pairs(all_trials)
    print(len(pairs))
    rows = []

    print(f"Classifying {len(pairs)} guess reasonings...\n")

    for p in tqdm(pairs, desc="Classifying guesses"):
        prompt = build_prompt(p["reasoning"], p["guessed_model"])
        result = client.classify(prompt)
        rows.append({
            **p,
            "strategies": ", ".join(result.get("Strategies", [])),
            "judge_explanation": result.get("Reasoning_behind_classification", "")
        })

    return pd.DataFrame(rows)

def run_classification_parallel(all_trials, api_key, max_workers=4):
    pairs = extract_pairs(all_trials)
    print(f"Classifying {len(pairs)} guess reasonings using {max_workers} threads...\n")

    batch_size = math.ceil(len(pairs) / max_workers)
    batches = [pairs[i:i + batch_size] for i in range(0, len(pairs), batch_size)]

    clients = [ORClient(api_key) for _ in range(len(batches))]

    results = []

    def classify_batch(batch, client):
        batch_results = []
        for p in batch:
            prompt = build_prompt(p["reasoning"], p["guessed_model"])
            result = client.classify(prompt)
            batch_results.append({
                **p,
                "strategies": ", ".join(result.get("Strategies", [])),
                "judge_explanation": result.get("Reasoning_behind_classification", "")
            })
        return batch_results

    with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = [
            executor.submit(classify_batch, batch, clients[i])
            for i, batch in enumerate(batches)
        ]

        results = []
        with tqdm(total=len(futures), desc="Classifying batches") as pbar:
            for f in concurrent.futures.as_completed(futures):
                results.extend(f.result())
                pbar.update(1)

    return pd.DataFrame(results)
````

````python
# classified_df = run_classification(data)
# classified_df.to_csv("guess_strategy_labels.csv", index=False)
# classified_df.head()
classified_df = run_classification_parallel(data_cues, OPENROUTER_API_KEY, max_workers=16)
classified_df.to_csv("figures_and_files/cue_utilization.csv", index=False)
````

````python
df = pd.read_csv("figures_and_files/cue_utilization.csv")
df.head()

def rem_num(text):
    match = re.search(r'[a-zA-Z]', text)
    if match:
        return text[match.start():]
    return text

# Expand the comma-separated strategy strings into lists
df["strategy_list"] = df["strategies"].fillna("").apply(lambda s: [rem_num(x).strip().lower() for x in s.split(",") if x.strip()])

collapsed = df.explode("strategy_list")
collapsed["strategy_list"] = collapsed["strategy_list"].replace({
    "Other": "Other/Unclear", 
    "Failed to parse": "Other/Unclear"
})

# Count occurrences per guessed model
model_strategy_counts = (
    collapsed.explode("strategy_list")
      .groupby(["guessed_model", "strategy_list"])
      .size()
      .reset_index(name="count")
      .sort_values(["guessed_model", "count"], ascending=[True, False])
)

model_strategy_counts.to_csv("figures_and_files/cue_utilization_processed.csv")
model_strategy_counts
````

````python
from collections import Counter

model_guess_counts = Counter()

for trial in data_cues:
    for key in ["agent_0_guess", "agent_1_guess"]:
        if key in trial:
            model_guess_counts[trial[key].lower()] += 1
            
df = pd.read_csv("figures_and_files/cue_utilization_processed_clean.csv")
df["guessed_model"] = df["guessed_model"].str.lower()

df["true_guess_count"] = df["guessed_model"].map(model_guess_counts)
df = df[df["true_guess_count"].notnull()]
df["percentage"] = 100 * df["count"] / df["true_guess_count"]
df["percentage"] = df["percentage"].round(2)

# Step 4: Reshape to top 3 cues per model in wide format
def format_top_cues(group):
    top = group.sort_values("percentage", ascending=False).head(3)
    return pd.Series({
        "Top Cue 1": f"{top.iloc[0]['strategy_list']} ({top.iloc[0]['percentage']}%)",
        "Top Cue 2": f"{top.iloc[1]['strategy_list']} ({top.iloc[1]['percentage']}%)",
        "Top Cue 3": f"{top.iloc[2]['strategy_list']} ({top.iloc[2]['percentage']}%)",
    })

rows = []
for model, group in df.groupby("guessed_model"):
    top = group.sort_values("percentage", ascending=False).head(3)
    rows.append({
        "guessed_model": model,
        "Top Cue 1": f"{top.iloc[0]['strategy_list']} ({top.iloc[0]['percentage']}%)",
        "Top Cue 2": f"{top.iloc[1]['strategy_list']} ({top.iloc[1]['percentage']}%)",
        "Top Cue 3": f"{top.iloc[2]['strategy_list']} ({top.iloc[2]['percentage']}%)",
    })

top_cues_corrected = pd.DataFrame(rows)
print(top_cues_corrected.to_latex(index=False))
````
